# Reproduce results from Clay-Kuba-Tripathi (arXiv:2607.10418)

This notebook reproduces the exact mean, variance and distribution of the number of correct guesses $X_n$ under the optimal complete-feedback strategy for the asymmetric single-shelf shuffle. The underlying mathematics is due to Clay, Kuba and Tripathi; this notebook is an independent reference implementation.

```python
import sys, os
sys.path.insert(0, os.path.join(os.path.dirname(__file__), '..', 'src'))

from shelfshuffle import (
    SingleShelfShuffle, OptimalCompleteFeedbackStrategy,
    simulate, expected_correct, variance_correct, theoretical_pmf, nu,
)
import matplotlib.pyplot as plt
```

## Exact mean and variance

For the symmetric shuffle ($p=1/2$) the expected score is $E[X_n]=3n/4$.

```python
n, p = 52, 0.5
print('E[X_n] =', expected_correct(n, p), '  (3n/4 =', 0.75*n, ')')
print('Var(X_n) =', variance_correct(n, p), '  (n/16 =', n/16, ')')

n, p = 52, 0.65
print('E[X_n] =', expected_correct(n, p))
print('Var(X_n) =', variance_correct(n, p))
```

## Monte-Carlo vs theory

```python
scores = simulate(OptimalCompleteFeedbackStrategy(), n, p, games=50000, seed=0)
emp_mean = sum(scores)/len(scores)
emp_var = sum((x-emp_mean)**2 for x in scores)/(len(scores)-1)
print(f'empirical mean {emp_mean:.3f}  vs  theoretical {expected_correct(n,p):.3f}')
print(f'empirical var  {emp_var:.3f}  vs  theoretical {variance_correct(n,p):.3f}')
```

## Exact distribution

```python
pmf = theoretical_pmf(n, p)
ks = [k for k, v in enumerate(pmf) if v > 1e-6]
plt.figure(figsize=(8, 4))
plt.bar(ks, [pmf[k] for k in ks])
plt.xlabel('correct guesses k')
plt.ylabel('P(X_n = k)')
plt.title(f'exact distribution, n={n}, p={p}')
plt.show()
```
